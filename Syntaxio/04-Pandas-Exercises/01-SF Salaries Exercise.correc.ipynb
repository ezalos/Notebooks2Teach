**How many job titles were represented by a single person in 2013? (i.e., job titles with only one occurrence in 2013?)**

```python
sal_2013 = sal.loc[sal["Year"] == 2013]
sal_2013
```

```python
all_titles_count = sal_2013["JobTitle"].value_counts()
all_titles_count
```

```python
all_titles_count.value_counts()
```

```python
series_of_unique_titles = all_titles_count == 1
series_of_unique_titles
```

```python
series_of_unique_titles.sum()
```

```python
sal_2013 = sal.loc[sal["Year"] == 2013]

list_of_all_job_titles = sal_2013["JobTitle"].unique().tolist()

list_of_all_job_titles
```

**How many people have the word "CHIEF" in their job title? (This is a bit tricky)**


```python
full_form = sal["JobTitle"].str.lower().str.contains("chief").sum()
full_form
```

```python
short_form = sal["JobTitle"].str.lower().str.contains("chf").sum()
short_form
```

```python
full_form + short_form
```

**Bonus: Is there a correlation between the length of the job title string and the salary?**
