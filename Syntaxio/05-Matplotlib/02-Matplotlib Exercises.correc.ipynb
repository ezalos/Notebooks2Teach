Follow the instructions to recreate the plots using this data:

## Data

```python
import numpy as np
x = np.arange(0,100)
y = x*2
z = x**2
```

**Import matplotlib.pyplot as plt and set %matplotlib inline if you are using Jupyter notebook. What command do you use if you are not using Jupyter notebook?**

```python
import matplotlib.pyplot as plt

%matplotlib inline

# plt.show()
```

## Exercise 1

Follow these steps:
* Create a figure object called fig using plt.figure()
* Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax.
* Plot (x, y) on these axes and set the labels and titles to match the plot below:

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("title")
```

## Exercise 2

Create a figure object and place two axes on it, ax1 and ax2. Located at [0,0,1,1] and [0.2,0.5,.2,.2] respectively.

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])
ax2 = fig.add_axes([0.2,0.5,.2,.2])
```

**Now plot (x, y) on both axes. And call your figure object to display it.**

```python
ax1.plot(x, y, c="r")
ax2.plot(x, y, c="r")
fig
```

## Exercise 3

Create the plot below by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,.4,.4]

```python
fig = plt.figure()
ax1 = fig.add_axes([0, 0, 1, 1])
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])
```

**Now use the arrays x, y and z to recreate the plot below. Note the x and y limits on the inset plot:**

```python
ax1.plot(x, z, c="b")
ax2.plot(x, y, c="b")

ax2.set_xlim(20, 22)
ax2.set_ylim(30, 50)

ax2.set_xlabel("x")
ax2.set_ylabel("y")

ax1.set_xlabel("x")
ax1.set_ylabel("z")

fig
```
