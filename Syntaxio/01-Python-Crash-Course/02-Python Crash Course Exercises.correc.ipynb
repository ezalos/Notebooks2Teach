```python
def findDog(st):
    return 'dog' in st
```

```python
findDog('Is there a dog here?')
```

```python
def findDog(st):
    for word in st.split():
        if 'dog' == word:
            return True
    return False
```

**Create a function that counts the number of times the word "dog" appears in a string.** 

Again, ignore special cases.


```python
def countDog(st):
    print(f"{st = }")

    sentence_as_list = st.split()
    print(f"{sentence_as_list = }")

    dog_word_list = [word for word in sentence_as_list if "dog" in word]
    print(f"{dog_word_list = }")

    count_dog = len(dog_word_list)
    print(f"{count_dog = }")

    return count_dog
```

```python
countDog('This dog runs faster than the other dog dude!')
```

```python
def countDog(st):
    return st.count("dog")
```

```python
countDog("This dog runs faster than the other dog dude!")
```

**You're driving a bit too fast, and a police officer stops you.**

Write a function to return one of 3 possible results: 
- "No Ticket", 
- "Small Ticket", 
- "Big Ticket". 

If your speed is 60 or less, the result is "No Ticket". 

If speed is between 61 and 80 inclusive, the result is "Small Ticket". 

If speed is 81 or more, the result is "Big Ticket". 

Unless it is your birthday (encoded as a boolean value in the parameters of the function) - on your birthday, your speed can be 5 km/h higher in all cases.


```python
def caught_speeding(speed: int, is_birthday: bool):
    if is_birthday:
        speed -= 5
    
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
```

```python
caught_speeding(81,True)
```

```python
caught_speeding(81,False)
```
