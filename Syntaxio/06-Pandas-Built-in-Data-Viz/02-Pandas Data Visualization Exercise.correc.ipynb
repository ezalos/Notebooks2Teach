# Pandas Data Visualization Exercise

This is a quick exercise to help you review the different plots we presented earlier. 

Use **df3** to replicate the following plots.

```python
import pandas as pd
import matplotlib.pyplot as plt
df3 = pd.read_csv('df3')
%matplotlib inline
```

```python
df3.info()
```

```python
df3.head()
```

**Recreate this scatter plot of b vs a. Note the color and size of the points. Also note the size of the figure. See if you can figure out how to stretch it in the same way. Don't forget your matplotlib lecture...**

```python

df3.plot.scatter(
	x='a',
	y='b',
	figsize=(12,3),
	c='red',
	edgecolor='black',
	s=50
)
```

**Create a histogram of column 'a'.**

```python
df3['a'].plot.hist(
	edgecolor='black',
	color='blue'
)
```

**These plots are correct, but they don't look very polished. Use style sheets to set the style to "ggplot" and redo the histogram from above. Also, find out how to add more bins to it.**

```python
plt.style.use('ggplot')
```

```python
df3['a'].plot.hist(
	edgecolor='black',
	bins=20
)
```

**Create a boxplot comparing columns a and b.**

```python
df3[['a', 'b']].plot.box()
```

**Create a kde plot of column 'd'**

```python
df3["d"].plot.kde()
```

**Discover how to increase the line width and make the line style dashed.**

```python
df3['d'].plot.kde(
	linewidth=5,
	linestyle=':'
)
```

**Create an area plot of all columns only for rows up to 30. (Hint: use .iloc).**

```python
df3.iloc[:30].plot.area(
	alpha=0.4
)
```
