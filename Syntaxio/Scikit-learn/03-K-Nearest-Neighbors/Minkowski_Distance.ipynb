```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
import random
import matplotlib.colors as mcolors


def perceived_brightness(hex_color):
    # Convert hex to RGB
    rgb = mcolors.to_rgb(hex_color)
    # Calculate perceived brightness
    return 0.299 * rgb[0] + 0.587 * rgb[1] + 0.114 * rgb[2]



def random_named_color(min_brightness=0.2, max_brightness=0.8):
    colors = list(mcolors.CSS4_COLORS.items())
    while True:
        name, hex_color = random.choice(colors)
        brightness = perceived_brightness(hex_color)
        if min_brightness <= brightness <= max_brightness:
            return name

```

```python
import matplotlib.pyplot as plt

HALF_PI = np.pi / 2
RADIUS = 1

def get_circle_points(p):
    xs_coords = []
    ys_coords = []
    for i in range(4):
        angles = np.linspace(0, HALF_PI, 1000)
        for angle in angles:
            x = np.cos(angle + HALF_PI * i)
            y = np.sin(angle + HALF_PI * i)
            scale = np.power(np.power(np.absolute(x), p) + np.power(np.absolute(y), p), 1 / p)
            
            x_coord = x * RADIUS / scale
            y_coord = y * RADIUS / scale
            
            if angle == 0 and i ==0:
                print(f"{p = } {angle = } {i = } {x_coord = }, {y_coord = }")
            xs_coords.append(x_coord)
            ys_coords.append(y_coord)
    return xs_coords, ys_coords


fig, ax = plt.subplots(figsize=(10, 10))

p_values = np.array([0.1, 0.25, 0.5, 0.75, 1.0, 1.5, 2.0, 5.0, 10.0, 1000.0])
p_values = np.array([0.1, .5, 1.0, 2.0, 5, 1000.0])
print(f"{p_values = }")

colors = plt.cm.winter(np.linspace(0, 1, len(p_values)))

for i, p in enumerate(p_values):
    xs_coords, ys_coords = get_circle_points(p)
    color = colors[i]
    color = random_named_color()
    print(f"{i = } {p = } {color = }")
    ax.plot(xs_coords, ys_coords, color=color, label=f"p={p:.1f}")

ax.legend()
ax.axis("equal")
ax.set_title("Minkowski Circles \n $ (\\sum_{i=1}^{n} |x_i - y_i|^p)^{1/p} = 1$")
```

```python
fig, ax = plt.subplots(1, 2)

NB_POINTS = 100
x = np.random.rand(NB_POINTS, 2)
y = np.random.rand(NB_POINTS, 2)

for i in range(NB_POINTS):
    color = random_named_color()
    ax[0].scatter(x[i][0], y[i][0], c=color)
    ax[0].scatter(x[i][1], y[i][1], c=color)
    ax[0].plot([x[i][0], x[i][1]], [y[i][0], y[i][1]], c=color)
ax[0].set_title("2D")

p_values = [1, 2, 4, 8, 16, 32, 64]
z_values = []
for p in p_values:
    delta_x = np.absolute(x[:, 0] - y[:, 0])
    delta_y = np.absolute(x[:, 1] - y[:, 1])
    z = ((delta_x ** p) + (delta_y ** p)) ** (1 / p)
    z_values.append(z)

ax[1].bar(p_values, z_values)
ax[1].set_title("Minkowski Distance")

plt.legend()
```

```python
fig, ax = plt.subplots(1, 2)

x_0, x_1 = np.random.rand(2)
y_0, y_1 = np.random.rand(2)

ax[0].scatter(x_0, y_0, c="r")
ax[0].scatter(x_1, y_1, c="b")
ax[0].set_title("2D")

p_values = [1, 2, 4, 8, 16, 32, 64]
z_values = []
for p in p_values:
    z = (np.absolute(x_0 - x_1) ** p + np.absolute(y_0 - y_1) ** p) ** (1 / p)
    z_values.append(z)

ax[1].bar(p_values, z_values)
ax[1].set_title("Minkowski Distance")

plt.legend()
```
