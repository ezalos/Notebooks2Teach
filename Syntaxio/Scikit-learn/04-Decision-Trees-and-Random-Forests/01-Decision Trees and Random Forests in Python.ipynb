# Decision Trees and Random Forests in Python

Decision trees are powerful and intuitive machine learning algorithms used for both classification and regression tasks. 

They work by recursively splitting the data based on features to create a tree-like structure of decision rules. Decision trees are particularly useful in scenarios where:
 
1. Interpretability is crucial: The decision-making process can be easily visualized and explained.
2. Non-linear relationships exist: They can capture complex interactions between features.
3. Mixed data types are present: They can handle both numerical and categorical data.
4. Feature importance needs to be determined: The tree structure reveals which features are most influential.
 
Common applications include:
 - Medical diagnosis
 - Customer churn prediction
 - Credit risk assessment
 - Fraud detection
 
In this notebook, we'll explore decision trees and their ensemble counterpart, random forests, using a medical dataset to predict the presence of kyphosis, a spinal condition.


## Import Libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
````

## Get the data

````python
df = pd.read_csv('kyphosis.csv')
````

````python
df.head()
````

````python
df.shape

# https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html
# https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.ADASYN.html
````

## EDA

We will simply check a simple pairplot for this small dataset.

````python
sns.pairplot(df,hue='Kyphosis',palette='Set1')
````

## Train-Test Split

Let's divide the data into a training set and a test set!

````python
from sklearn.model_selection import train_test_split
````

````python
X = df.drop('Kyphosis',axis=1)
y = df['Kyphosis']
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
````

## Decision Trees

We will start simply by training a single decision tree.

````python
from sklearn.tree import DecisionTreeClassifier
````

````python
dtree = DecisionTreeClassifier()
````

````python
dtree.fit(X_train,y_train)
````

## Prediction and Evaluation

Let's evaluate our decision tree.

````python
predictions = dtree.predict(X_test)
````

````python
from sklearn.metrics import classification_report,confusion_matrix
````

````python
print(classification_report(y_test,predictions))
````

Here's an ASCII representation of the confusion matrix:
```
           Predicted
           0   |   1
         -------------
 Actual 0 | TN  |  FP
        1 | FN  |  TP
```
 Where:
- TP = True Positives
- TN = True Negatives
- FP = False Positives
- FN = False Negatives


````python
print(confusion_matrix(y_test,predictions))
````

## Tree Visualization

Scikit-learn actually has built-in visualization capabilities for decision trees. 

You won't use it often, but here's an example of what it looks like and the code to run it:

````python
from sklearn import tree

features = list(df.columns[1:])
features
tree.plot_tree(dtree, feature_names=features, class_names=['absent','present'], filled=True)

None
````

## Random Forests

Let's now compare the decision tree model to a random forest.

````python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(n_estimators=100)
rfc.fit(X_train, y_train)
````

````python
rfc_pred = rfc.predict(X_test)
````

Here's an ASCII representation of the confusion matrix:
```
           Predicted
           0   |   1
         -------------
 Actual 0 | TN  |  FP
        1 | FN  |  TP
```
 Where:
- TP = True Positives
- TN = True Negatives
- FP = False Positives
- FN = False Negatives


````python
print(confusion_matrix(y_test,rfc_pred))
````

````python
print(classification_report(y_test,rfc_pred))
````

## K-Fold

With such a small sample size, it's hard to have confidence in our results, let's use K-Fold

````python
# Let's use K-Fold to train our model
from sklearn.model_selection import KFold
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np

kfold = KFold(n_splits=10, shuffle=True, random_state=42)

all_predictions = []
all_true_labels = []

for train_index, test_index in kfold.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    rfc = RandomForestClassifier(n_estimators=100)
    rfc.fit(X_train, y_train)
    rfc_pred = rfc.predict(X_test)
    
    all_predictions.extend(rfc_pred)
    all_true_labels.extend(y_test)

# Merge results and compute overall performance
print("Overall Confusion Matrix:")
print(confusion_matrix(all_true_labels, all_predictions))
print("\nOverall Classification Report:")
print(classification_report(all_true_labels, all_predictions))
````
