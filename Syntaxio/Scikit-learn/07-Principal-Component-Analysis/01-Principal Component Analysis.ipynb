## PCA Review

## Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
```

## The Data

Let's work again with the cancer dataset, as it has many features.

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
cancer.keys()
```

```python
print(cancer['DESCR'])
```

```python
df = pd.DataFrame(cancer['data'],columns=cancer['feature_names'])
#(['DESCR', 'data', 'feature_names', 'target_names', 'target'])
```

```python
df.head()
```

## PCA Visualization

As we have already noticed, it is difficult to visualize high-dimensional data. We can use PCA to find the first two principal components and visualize the data in this new two-dimensional space, with a single scatter plot. However, before doing this, we will need to scale our data so that each feature has unit variance.

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

PCA with Scikit-Learn uses a process very similar to other preprocessing functions provided by SciKit-Learn. We instantiate a PCA object, find the principal components using the fit method, and then apply rotation and dimensionality reduction by calling transform().

We can also specify the number of components we want to retain when creating the PCA object.

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca.fit(scaled_data)
```

We can now transform this data into its first 2 principal components.

```python
x_pca = pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
x_pca.shape
```

Great! We've reduced 30 dimensions to just 2! Let's plot these two dimensions!

```python
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0],x_pca[:,1],c=cancer['target'],cmap='plasma')
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```

Clearly, using these two components, we can easily separate these two classes.

## Interpretation of Components

Unfortunately, this great power of dimensionality reduction comes at the cost of being able to easily understand what these components represent.

The components correspond to combinations of the original features, and the components themselves are stored as an attribute of the fitted PCA object:

```python
pca.components_
```

In this numpy matrix array, each row represents a principal component and each column refers to the original features. We can visualize this relationship with a heatmap:

```python
df_comp = pd.DataFrame(np.absolute(pca.components_),columns=cancer['feature_names'])
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(df_comp,cmap='plasma',)
```

This heatmap and color bar essentially represent the correlation between the different features and the principal component itself.
