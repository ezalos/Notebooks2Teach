# K-Means Clustering with Python

## Method Used

K-Means Clustering is an unsupervised learning algorithm that attempts to group similar data points together. 

Unsupervised learning means there are no outcome variables to predict, and the algorithm just tries to find patterns in the data. 

In K-Means clustering, we need to specify the number of clusters we want the data to be grouped into. The algorithm randomly assigns each observation to a cluster and finds the centroid of each cluster. Then, the algorithm iterates through two steps:

- Reassign data points to the cluster whose centroid is closest.
- Calculate the new centroid of each cluster.

These two steps are repeated until the within-cluster variation cannot be reduced any further.

The within-cluster variation is calculated as the sum of the Euclidean distance between the data points and their respective cluster centroids.

## Import Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

## Create Data

```python
from sklearn.datasets import make_blobs
```

```python
# Create Data
data = make_blobs(n_samples=200, n_features=2, 
                           centers=4, cluster_std=1.8,random_state=101)
```

## Visualize the data

```python
plt.scatter(data[0][:,0],data[0][:,1],c=data[1],cmap='rainbow')
```

## Creating the clusters

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4)

# K-means Hyper Parameters:
# - Init method of k-means++: https://theory.stanford.edu/~sergei/papers/kMeansPP-soda.pdf

# For automatic choosing of hyper parameter K, there exists the X-Means algorithm:
# - Describe here: https://www.cs.cmu.edu/~dpelleg/download/xmeans.pdf
# - Implemented here: https://github.com/alex000kim/XMeans/blob/master/xmeans.py
```

```python
kmeans.fit(data[0])
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_
```

```python
f, (ax1, ax2) = plt.subplots(1, 2, sharey=True,figsize=(10,6))
ax1.set_title('K Means')
ax1.scatter(data[0][:, 0], data[0][:, 1], c=kmeans.labels_, cmap="rainbow")
ax1.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c="black", s=100, alpha=0.75)
ax2.set_title("Original")
ax2.scatter(data[0][:, 0], data[0][:, 1], c=data[1], cmap="rainbow")
```

Note: The colors have no meaningful reference between the two plots.
