# Steps

1. Transform business problem to ml problem
2. Define error metric
3. Training loop
   1. Get to know data
      1. Pair plot -> ML intuition 
      2. Classification:
         1. Are class imbalance ?
      3. Regression:
         1. Correlation with target variable ?
      4. Target variable 
   2. Clean
      1. Impute missing data or drop
      2. Convert variables to usable types
      3. Remove outliers
      4. Remove overfited variables
   3. Feature engineering
      1. Feature extraction
      2. Feature engineering
      3. Feature selection
      4. Study if features validates our model assumptions
   4. Save train / val / test sets
   5. Train models
      1. Train normalizer on data
      2. Select 2,3 models in the ML category
      3. Hyper parameter tuning
      4. Enventually classification threshold
   6. Evaluate:
      1. Single number metric
         1. Human / Train / (Val & Test) error -> helps know if underfitting, overfitting, or a possibly a problem not sure it can be solved by ML.
      2. Confusion matrix
      3. Error analysis
         1. If I was an expert, what information would I be missing to make better predictions ?
      4. Define bottlenecks
