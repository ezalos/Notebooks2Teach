```python
import torch
import torch.nn as nn
import numpy as np
import sklearn
from sklearn import datasets
from sklearn.preprocessing import StandardScaler  # for feature scaling
from sklearn.model_selection import train_test_split  # for train/test split
```

```python
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
sklearn.utils.validation.check_random_state(SEED)
```

# Generate the data

```python
from sklearn.datasets import make_classification

# Generate the data
X, y = make_classification(
	n_samples=100, 
	n_features=5,
	n_informative=2, 
	n_redundant=2, 
	n_clusters_per_class=1,
	weights=[0.8, 0.2]
)

# Add some noise to the data
X += np.random.randn(100, 5)

X.shape, y.shape
```

```python
X.dtype, y.dtype
```

```python
X.mean(), X.std()
```

## Prepare data

Set the variables `n_samples` and `n_features` to the correct values.

```python
n_samples, n_features = X.shape
print(f"number of samples: {n_samples}, number of features: {n_features}")
```

Split the data into training and test sets, with 20% of the data being used for testing.

Use the seed 1234 for the random state.

```python
# split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1234
)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

Use a StandardScaler to scale the data.

```python
# scale data
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

Convert the numpy arrays data X_train, X_test, y_train, y_test to PyTorch tensors.

```python
# convert to tensors
X_train = torch.from_numpy(X_train.astype(np.float32))
X_test = torch.from_numpy(X_test.astype(np.float32))
y_train = torch.from_numpy(y_train.astype(np.float32))
y_test = torch.from_numpy(y_test.astype(np.float32))

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

Make sure to reshape the y tensors, so they have a feature dimension.

```python
# reshape y tensors, so they have a feature dimension

y_train = y_train.reshape((y_train.shape[0], 1))
y_test = y_test.view(y_test.shape[0], 1)

y_train.shape, y_test.shape
```

## Our model

You already have the model definition ready to be used 

```python
# f = sigmoid(WX + b)
class LogisticRegression(nn.Module):

    def __init__(self, n_input_features):
        super().__init__()
        self.linear = nn.Linear(n_input_features, 1) # linear = [w1, w2, ..., wn, b]

    def forward(self, x):
        z = self.linear(x) # z = x_1*w_1 + x_2*w_2 + ... + x_n*w_n + b
        y_predicted = torch.sigmoid(z)
        return y_predicted


model = LogisticRegression(n_features)
model
```

Define the optimizer

We will use the SGD optimizer, with a learning rate of 0.1.

```python
# Loss and optimizer
learning_rate = 0.1
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
```

Define the loss_function, we will use the BCELoss

```python
loss_function = nn.BCELoss()
```

# Using the GPU

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)
```

# Training loop

fill in the missing code to train the model

```python
num_epochs = 100
```

```python
# training loop
for epoch in range(num_epochs):

    # forward pass and loss
    y_predicted = model.forward(X_train)
    loss = loss_function(y_predicted, y_train)

    # backward pass
    loss.backward()

    # updates
    optimizer.step()

    # zero gradients
    optimizer.zero_grad()
    
    if (epoch + 1) % 10 == 0:
        print(f"epoch: {epoch+1}, loss = {loss.item():.4f}")
```

Evalutate your model on the test set

```python
with torch.no_grad():
    y_predicted = model.forward(X_test)
    y_predicted_cls = y_predicted.round()  # round off to nearest class
    acc = y_predicted_cls.eq(y_test).sum() / float(y_test.shape[0])  # accuracy
    print(f"accuracy test_set = {acc:.4f}")
```
