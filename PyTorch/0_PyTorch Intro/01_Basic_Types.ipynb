## Tensors

Tensors are similar to NumPy arrays but can also be used on GPUs and CPUs


```python
# Tensors are similar to NumPy arrays 
# but can also be used on GPUs (and other accelerators)

# Create a tensor from a Python list
tensor = torch.tensor([1, 2, 3, 4, 5])

tensor
```

```python
tensor.shape
```

```python
tensor.dtype
```

```python
# We can also convert a numpy array to a tensor
numpy_array = np.array([1, 2, 3, 4, 5])
print(f"{numpy_array = }")
tensor_from_numpy = torch.from_numpy(numpy_array)
tensor_from_numpy
```

```python
# We can also convert a tensor to a list
tensor_to_list = tensor_from_numpy.tolist()
tensor_to_list
```

```python
tensor_to_numpy = tensor_from_numpy.numpy()
tensor_to_numpy
```

We will retrieve many familiar functions from NumPy


```python
t_ones = torch.ones((3, 4))
t_ones
```

```python
t_zeros = torch.zeros((3, 4))
t_zeros
```

```python
t_rand = torch.rand((3, 4))
t_rand
```

```python
# Like with numpy we can reshape tensors
t_rand.reshape(2, 6)
```

```python
t_rand.reshape(-1, 2)
```

# Using the GPU

```python
# Let's check if cuda is available:
if torch.cuda.is_available():
    print(f"Cuda is available!")
    device = torch.device("cuda")
else:
    print(f"CUDA is not available! Let's use the CPU")
    device = torch.device("cpu")

# This device object will be used to specify where a tensor should be stored.
device
```

```python
# We can send data on the gpu this way:
tensor = torch.ones((3, 4))

tensor_on_gpu = tensor.to(device)
tensor_on_gpu
```

```python
tensor_on_gpu = tensor.to("cuda")
tensor_on_gpu
```

```python
# We can also do some operations on the gpu
tensor_on_gpu + tensor_on_gpu
```

```python
# But be careful, only tensors on the same device can be used in operations

print(f"{tensor_on_gpu = }")
print(f"{t_ones = }")

# tensor_on_gpu + t_ones # /!\ This will raise an error
```

```python
# tensor_on_gpu.numpy()
```

```python
tensor_on_gpu.to("cpu").numpy()
```

```python
tensor_on_gpu.cpu().numpy()
```

```python
# Once we are done with the gpu operations we can send the tensor back to the cpu
tensor_on_cpu = tensor_on_gpu.to("cpu")
tensor_on_cpu
```

```python
# Or even to the numpy
numpy_array = tensor_on_cpu.numpy()
numpy_array
```
